# EXPLORATORY DATA ANALYSIS:

```python
# Monthly revenue trend
monthly = df.groupby(df['order_date'].dt.to_period('M'))['total_revenue'].sum()
monthly.plot(figsize=(10,5), title="Monthly Revenue Trend")
plt.ylabel("Revenue")
plt.show()
```

## BUISNESS INSIGHTS:
##### 1.Highest monthly revenue was achieved in January 2023 (~1.46M) could be due to New Year sales, festive demand,or high customer purchases.
##### 2.Lowest monthly revenue occurred in February 2023 (~1.24M).
##### 3.Revenue shows high month-to-month volatility, indicating seasonal or promotional effects.
##### 4.The business demonstrates strong recovery after revenue declines.
##### 5.Revenue remains relatively stable within the 1.30M–1.45M range,showing a consistent customer base.

```python
# Revenue by category
cat_rev = df.groupby('product_category')['total_revenue'].sum().sort_values(ascending=False)
sns.barplot(x=cat_rev.values, y=cat_rev.index, color = 'pink')
plt.title("Revenue by Category")
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Beauty is the highest revenue-generating category.
##### 2.Sports contributes the lowest revenue among all categories.
##### 3.Revenue is evenly distributed across all product categories.
##### 4.Fashion, Books, and Home & Kitchen are strong contributors to total sales.
##### 5.Electronics maintains consistent revenue performance.

```python
# RATING BY CATEGORY
plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='product_category', y='rating',marker='o',color = 'orange')
plt.title("Rating  by Category")
plt.xlabel('CATEGORY')
plt.ylabel('RATING')
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Books have the highest average customer rating, indicating the best customer satisfaction among all categories.
##### 2.Beauty has the lowest rating, suggesting improvements are needed in product quality or customer experience.
##### 3.Customer ratings are consistent across all categories, averaging around 3/5, indicating moderate satisfaction.
##### 4.Differences between category ratings are small, so no category significantly outperforms the others.
##### 5.The business should focus on improving product quality, delivery, and customer service rather than relying only on category-specific strategies.


```python
# Revenue by region
plt.figure(figsize=(12,6))
region_rev = df.groupby('customer_region')['total_revenue'].sum().sort_values(ascending=False)
sns.barplot(x=region_rev.values, y=region_rev.index, palette="viridis", hue = region_rev.values, legend = False)
plt.title("Revenue by Region")
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Middle East is the highest revenue-generating region.
##### 2.Europe records the lowest revenue among all regions.
##### 3.Revenue is well balanced across all geographic regions.
##### 4.North America consistently delivers strong sales performance.
##### 5.Asia has significant growth potential due to its large customer base.


```python
# Discount vs category relationship
plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='product_category',y='discount_percent',marker='o', c= 'g')
plt.title("Discount % by Category")
plt.xlabel('Discount')
plt.ylabel('Category')
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Sports offers the highest average discount, while Electronics has the lowest.
##### 2.Discounts are nearly uniform (around 13%) across all categories.
##### 3.The company follows a consistent pricing and promotion strategy across product categories.
##### 4.Uniform discounts may not maximize profits because different categories have different demand levels.
##### 5.A category-specific discount strategy could improve both sales and profitability.


```python
# Payment method popularity
plt.figure(figsize=(15,15))
df['payment_method'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))
plt.title("Orders by Payment Method"); plt.ylabel("")
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Orders are almost equally distributed among all payment methods.
##### 2.UPI and Wallet are the most preferred payment methods, indicating strong digital payment adoption.
##### 3.Credit Card usage is slightly lower than other payment options.
##### 4.Cash on Delivery (COD) still accounts for a significant share of orders.
##### 5.The company should continue supporting all payment methods while encouraging digital payments through rewards and cashback offers.


```python
# Discount vs rating relationship
plt.figure(figsize=(12,6))

sns.lineplot(data=df, x='discount_percent', y='rating',marker='o')
plt.title("Discount % vs Rating")
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Increasing discounts does not significantly improve customer ratings.
##### 2.Customer ratings remain stable around 3/5 regardless of discount percentage.
##### 3.Higher discounts produce only a slight improvement in customer satisfaction.
##### 4.Customer experience depends more on product quality and service than discounts.
##### 5.Discounts should be used as marketing tools rather than the primary strategy for improving customer satisfaction.


```python
# Correlation heatmap
plt.figure(figsize=(12,6))
sns.heatmap(df[['price','discount_percent','quantity_sold','rating','review_count','total_revenue']].corr(),annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Price has the strongest positive relationship with Total Revenue (0.71).
##### 2.Quantity Sold is the second strongest driver of revenue (0.59).
##### 3.Discount Percentage has a slight negative relationship with revenue (-0.14).
##### 4.Ratings and Review Count have almost no direct relationship with revenue.
##### 5.Revenue growth should focus on pricing strategy and increasing sales volume, while discounts should be applied strategically.


```python
# ORDER BY REGION
plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='customer_region', y='quantity_sold',marker='o', color='purple')
plt.title("Orders by Region")
plt.xlabel('REGION')
plt.ylabel('ORDER')
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.The Middle East records the highest average number of orders.
##### 2.Asia has the lowest average order volume among the regions.
##### 3.North America and Europe show stable and similar order performance.
##### 4.Regional differences in orders are relatively small, indicating balanced global demand.
##### 5.The company should expand marketing efforts in the Middle East while improving customer engagement and sales strategies in Asia.


```python
# Revenue by discount
plt.figure(figsize=(12,6))
dis_rev = df.groupby('discount_percent')['total_revenue'].sum().sort_values(ascending=False)
sns.barplot(x=dis_rev.index, y=dis_rev.values,palette='Spectral', hue = dis_rev.index, legend = False)
plt.title("Revenue by discount")
plt.xlabel('discount %')
plt.ylabel('Revenue')
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Revenue is highest with little or no discount. The 0% and 5% discount levels generate the highest revenue (around 6.2 million).
##### 2.Revenue declines as discount percentage increases. There is a clear downward trend in revenue from 10% to 30% discounts.
##### 3.High discounts reduce profitability. The 30% discount generates the lowest revenue (around 4.4 million).
##### 4.Moderate discounts are more effective than heavy discounts. Discounts of 5–10% maintain relatively high revenue while still providing promotional value, making them more sustainable than larger discounts.
##### 5.Optimize discount strategy for maximum revenue. Instead of offering large discounts across all products, the business should use targeted and seasonal promotions, personalized offers, or discounts on slow-moving inventory to maximize revenue while protecting profit margins.

```python
# Revenue by weekday
plt.figure(figsize=(12,6))
week_rev = df.groupby('weekday')['total_revenue'].sum().sort_values(ascending=False)
sns.barplot(x=week_rev.index, y=week_rev.values,palette='mako', hue = week_rev.index, legend = False
        )
plt.title("Revenue by weekday")
plt.xlabel('Weekday')
plt.ylabel('Revenue')
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Friday generates the highest revenue, making it the best day for sales and promotions.
##### 2.Monday records the lowest revenue, indicating an opportunity to boost sales with targeted offers.
##### 3.Weekend (Saturday & Sunday) sales are consistently strong, showing higher customer purchasing activity.
##### 4.Revenue remains fairly stable throughout the week, indicating consistent customer demand.
##### 5.Targeted weekday promotions (Monday–Thursday) can help increase revenue and balance sales across the week.

```python
# Order by weekday
plt.figure(figsize=(12,6))
week_ord = df.groupby('weekday')['quantity_sold'].sum().sort_values(ascending=False)
sns.barplot(x=week_ord.index, y=week_ord.values,palette="viridis", hue = week_ord.index, legend = False)
plt.title("Order by weekday")
plt.xlabel('Weekday')
plt.ylabel('Order')
plt.show()
```

# BUISNESS INSIGHTS:
##### 1.Friday has the highest number of orders, making it the busiest sales day of the week.
##### 2.Wednesday records the lowest number of orders, indicating lower customer activity.
##### 3.Weekend (Saturday and Sunday) order volumes remain strong, reflecting increased shopping during weekends.
##### 4.Order volumes are fairly consistent across all weekdays, showing stable customer demand throughout the week.
##### 5.Weekday-specific promotions on Monday to Wednesday can help increase orders and balance demand across the week.

# CONCLUSION

#### Based on the analysis of all 12 graphs and charts from the Amazon Data Analytics Project, the following key business conclusions can be drawn:

##### 1.The business demonstrates stable overall performance, with revenue remaining consistent despite monthly fluctuations caused by seasonal demand and promotional campaigns.
##### 2.Product categories contribute almost equally to total revenue, with Beauty emerging as the top-performing category, while Sports offers an opportunity for revenue improvement.
##### 3.Customer satisfaction is moderate across all categories (around 3/5), indicating that improving product quality, delivery services, and customer support should be a key business priority.
##### 4.Pricing and sales volume are the primary drivers of revenue, as price and quantity sold show the strongest positive relationship with total revenue.
##### 5.Heavy discounting does not improve business performance significantly. Higher discounts reduce revenue and have minimal impact on customer ratings, suggesting that discounts should be used strategically rather than frequently.
##### 6.A personalized and category-specific discount strategy would be more effective than offering uniform discounts across all product categories, helping maximize both sales and profit margins.
##### 7.Digital payment methods are widely accepted, with UPI and Wallet being the most preferred payment options, while maintaining multiple payment methods enhances customer convenience.
##### 8.Regional business performance is well balanced, with the Middle East showing the highest revenue and order volume, making it a strong market for future expansion, while Asia presents significant growth potential.
##### 9.Customer purchasing behavior is strongest on Fridays and weekends, making these periods ideal for launching marketing campaigns, promotional offers, and new product releases.
##### 10.Weekday sales and order volumes can be improved through targeted promotions, loyalty programs, and personalized offers, especially from Monday to Wednesday.
##### 11.Customer ratings and review counts have little direct impact on revenue, indicating that while customer satisfaction is important for long-term loyalty, revenue growth depends more on effective pricing strategies and increased sales volume.
##### 12.Overall, the business has a strong and diversified foundation, with balanced category performance, stable regional demand, and multiple payment options.Future growth can be achieved by optimizing pricing, improving customer experience, implementing data-driven promotional strategies, expanding in high-performing regions, and leveraging customer insights for informed business decisions.
